```python
import sqlite3

# 1. Read the SQL file
with open('setup.sql', 'r') as f:
    sql_query = f.read()

# 2. Connect to a database (creates it if it doesn't exist)
conn = sqlite3.connect('my_database.db')
cursor = conn.cursor()

# 3. Execute the script
# Note: use executescript() if your file contains multiple statements
cursor.executescript(sql_query)

conn.commit()
conn.close()
```
