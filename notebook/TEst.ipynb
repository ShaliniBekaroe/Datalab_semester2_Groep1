```python
from pathlib import Path
import sqlite3

# Define the directory
base_path = Path(r'C:\Users\sasha\Documents\GitHub\Datalab_semester2_Groep1\notebook\Test')
file_to_open = base_path / 'your_file.sql' # This joins them automatically

if file_to_open.exists():
    sql_script = file_to_open.read_text()
    
    # Execute
    conn = sqlite3.connect('database.db')
    conn.executescript(sql_script)
    conn.close()
    print(f"Successfully read: {file_to_open.name}")
else:
    print("File not found! Check the filename.")
```
